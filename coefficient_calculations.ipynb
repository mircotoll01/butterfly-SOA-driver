```python
# ------------------- Punti Misurati --------------------------
import numpy as np

# Temperature (Kelvin)
T_1 = 273.15  # 0°
T_2 = 303.15  # 30°
T_3 = 323.45  # 50.3°

# Resistenze (Ohm)
R_1 = 32435
R_2 = 7803
R_3 = 3456
```

```python
# ------------------- logaritmi vari e altro -----------------
L_1 = np.log(R_1)
L_2 = np.log(R_2)
L_3 = np.log(R_3)

Y_1 = 1/T_1
Y_2 = 1/T_2
Y_3 = 1/T_3

gamma_2 = (Y_2 - Y_1)/(L_2 - L_1)
gamma_3 = (Y_3 - Y_1)/(L_3 - L_1)

print("L_1", L_1, "\n")
print("L_2", L_2, "\n")
print("L_3", L_3, "\n")

print("Y_1", Y_1, "\n")
print("Y_2", Y_2, "\n")
print("Y_3", Y_3, "\n")

print("gamma_2", gamma_2, "\n")
print("gamma_3", gamma_3, "\n")
```

```python
# ---------------- calcolo coefficienti------------------------

C = ((gamma_3-gamma_2)/(L_3-L_2))*((L_1+L_2+L_3)**(-1))
B = gamma_2 - C*(L_1**2 + L_1*L_2 + L_2**2)
A = Y_1 - (B + (L_1**2)*C)*L_1

print("A = ", A, "\n")
print("B = ", B, "\n")
print("C = ", C, "\n")
```

```python
# ----------------- Conversione tensione-temperatura -----------
# partitore di tensione con 10Kohm ed NTC 10Kohm a temperatura ambiente

# Valori variabili (Volt)
V_CC  = 1.5   
V_NTC = 1.3

# calcoli
V_R = V_CC - V_NTC
I_R = V_R / 10000

R_NTC = V_NTC / I_R

print("R_NTC = ", R_NTC, "\n")

Y = A + B*(np.log(R_NTC)) + C*((np.log(R_NTC))**3)
T = 1/Y

print("T = ", T - 273.15, "°C\n")
```
